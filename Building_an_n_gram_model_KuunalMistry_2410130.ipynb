```python
import nltk
from collections import defaultdict
from nltk import trigrams
from nltk.tokenize import word_tokenize

# Download necessary NLTK resource packages
nltk.download('punkt')
nltk.download('punkt_tab')

# 1. Sample training corpus
corpus = [
    "This is a simple example of an n-gram model.",
    "N-gram models predict the probability of a word.",
    "This example demonstrates how n-gram language models work.",
]

# 2. Tokenize the corpus into sentences of words
tokenized_corpus = [word_tokenize(sent.lower()) for sent in corpus]

# 3. Initialize the model structure
# Structure: model[(word1, word2)][word3] = count
model = defaultdict(lambda: defaultdict(int))

# 4. Count frequencies of trigrams
for sentence in tokenized_corpus:
    # Pad left and right if needed, or just iterate over standard trigrams
    for w1, w2, w3 in trigrams(
        sentence, pad_right=True, pad_left=True
    ):
        model[(w1, w2)][w3] += 1

# 5. Convert counts to conditional probabilities P(w3 | w1, w2)
probability_model = defaultdict(lambda: defaultdict(float))
for w1_w2, w3_counts in model.items():
    total_count = float(sum(w3_counts.values()))
    for w3, count in w3_counts.items():
        probability_model[w1_w2][w3] = count / total_count

# 6. Test prediction: Predict the next word given two preceding words
context = ("this", "is")
print(f"Predictions for context {context}:")
for next_word, prob in sorted(
    probability_model[context].items(), key=lambda x: x[1], reverse=True
):
    print(f" Word: '{next_word}' (Probability: {prob:.2f})")
```
