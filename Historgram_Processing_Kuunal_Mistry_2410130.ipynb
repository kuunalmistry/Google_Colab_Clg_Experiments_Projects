```python
import cv2
```

```python
img = cv2.imread("/content/lowcontrast.jpg",0)
```

```python
cv2_imshow(img)
```

```python
hist8=cv2.calcHist([img],[0],None,[256],[0,256])
```

```python
plt.plot(hist8)
plt.xlabel("pixel value")
plt.ylabel("frequency")
plt.show()
```

```python
s_max=255
s_min=0
```

```python
r_min=np.min(img)
r_max=np.max(img)
```

```python
r_min
```

```python
r_max
```

```python
op_img=(s_max/(r_max-r_min))*(img-r_min)+s_min
```

```python
cv2_imshow(op_img)
```

```python
op_img=op_img.astype(np.uint8)
```

```python
hist9=cv2.calcHist([op_img],[0],None,[256],[0,256])
```

```python
plt.plot(hist9)
plt.xlabel("pixel value")
plt.ylabel("frequency")
plt.show()
```

```python
res=np.hstack((img,op_img))
```

```python
cv2_imshow(res)
```

```python
plt.subplot(2,2,1)
plt.plot(hist8)
plt.xlabel("pixel value")
plt.ylabel("frequency")
plt.title("Original Image")

plt.subplot(2,2,2)
plt.plot(hist9)
plt.xlabel("pixel value")
plt.ylabel("frequency")
plt.title("Equalized Image")
```

```python
hist10=cv2.calcHist([img],[0],None,[256],[0,256])
```

```python
plt.plot(hist10)
plt.title("Before Equilization")
plt.xlabel("pixel value")
plt.ylabel("frequency")
plt.show()
```

```python
img1=cv2.equalizeHist(img)
```

```python
cv2_imshow(img1)
```

```python
res=np.hstack((img,img1))
```

```python
cv2_imshow(res)
```

```python
hist11=cv2.calcHist([img1],[0],None,[256],[0,256])
```

```python
plt.plot(hist11)
plt.title("After Equilization")
plt.xlabel("pixel value")
plt.ylabel("frequency")
plt.show
```

```python
plt.subplot(2,2,1)
plt.plot(hist10)
plt.xlabel("pixel value")
plt.ylabel("frequency")
plt.title("Before Equilization")

plt.subplot(2,2,2)
plt.plot(hist11)
plt.xlabel("pixel value")
plt.ylabel("frequency")
plt.title("After Equilization")
```
