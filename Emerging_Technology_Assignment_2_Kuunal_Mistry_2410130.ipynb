```python
import sys

def dijkstra(graph, source):
    n = len(graph)

    dist = [sys.maxsize] * n
    visited = [False] * n

    dist[source] = 0

    for _ in range(n - 1):
        u = -1
        min_dist = sys.maxsize

        # Find vertex with minimum distance
        for i in range(n):
            if not visited[i] and dist[i] < min_dist:
                min_dist = dist[i]
                u = i

        # If no reachable unvisited node, break
        if u == -1:
            break

        visited[u] = True

        # Update adjacent vertices
        for v in range(n):
            if graph[u][v] != 0 and \
               not visited[v] and \
               dist[u] + graph[u][v] < dist[v]:
                dist[v] = dist[u] + graph[u][v]

    print("Shortest distances from A:")
    for i in range(n):
        print(f"{chr(ord('A') + i)} = {dist[i]}")

# Example usage:
# A=0, B=1, C=2, D=3
graph = [
    [0, 4, 2, 0],  # A
    [0, 0, 2, 5],  # B
    [0, 0, 0, 1],  # C
    [0, 0, 0, 0]   # D
]

dijkstra(graph, 0)
```
