```python
import cv2
import numpy as np
from skimage.util import random_noise
from google.colab.patches import cv2_imshow
```

```python
img = cv2.imread("/content/opencv Logo.png")
```

```python
gaussian = random_noise(img, mode='gaussian', mean = 0, var = 0.7)
gaussian = np.array(255*gaussian, dtype = 'uint8')
print("Gaussian Noise")
cv2_imshow(gaussian)
```

```python
salt_peper = random_noise(img, mode='s&p', amount = 0.01)
salt_peper = np.array(255*salt_peper, dtype = 'uint8')
print("Salt and Peper Noise")
cv2_imshow(salt_peper)
```

```python
poisson = random_noise(img, mode='poisson')
poisson = np.array(255 * poisson, dtype = 'uint8')
print("Poisson Noise")
cv2_imshow(poisson)
```

```python
img=cv2.imread("/content/SaltandPepper.png", 0)
```

```python
m, n = img.shape
```

```python
print(m)
```

```python
print(n)
```

```python
mask = np.ones([3, 3], dtype= int)
mask = mask / 9
```

```python
img_new = np.zeros([m, n])
```

```python
for i in range(1, m-1):
  for j in range(1, n-1):
    temp = img[i-1, j-1]*mask[0, 0]+img[i-1, j]*mask[0, 1]+img[i-1, j+1]*mask[0, 2]+img[i, j-1]*mask[1, 1]+img[i, j+1]*mask[1, 2]+img[i+1, j-1]*mask[2, 0]+img[i+1, j]*mask[2, 1]+img[i+1, j+1]*mask[2, 2]

    img_new[i, j]=temp
```

```python
img_new = img_new.astype(np.uint8)
cv2_imshow(img_new)
```

```python
imgblur = cv2.medianBlur(img, 9)
cv2_imshow(imgblur)
```

```python
imgblur1 = cv2.medianBlur(img, 11)
cv2_imshow(imgblur1)
```

```python
reels = np.hstack((img, imgblur,imgblur1))
cv2_imshow(reels)
```
