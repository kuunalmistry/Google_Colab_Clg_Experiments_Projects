```python
import cv2
```

```python
import numpy as np
```

```python
from google.colab.patches import cv2_imshow
```

```python
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('/content/1.jpg',0)
```

```python
point = np.array([[-1,-1,-1], [-1,8,-1], [-1, -1, -1]], dtype = np.float32)
```

```python
out = cv2.filter2D(img, -1, point)
```

```python
cv2_imshow(out)
```

```python
H = np.array([[-1,-1,-1], [2,2,2],[-1,-1,-1 ]], dtype = np.float32)
```

```python
h_out = cv2.filter2D(img, -1, H)
```

```python
cv2_imshow(h_out)
```

```python
V=np.array ([[-1,2,-1], [-1,2,-1], [-1,2,-1]], dtype = np.float32)
```

```python
v_out = cv2.filter2D(img, -1, V)
cv2_imshow(v_out)
```

```python
M45 = np.array([[-1,-1,2], [-1,2,-1], [2,-1,-1]], dtype = np.float32)
```

```python
M45_out = cv2.filter2D(img, -1, M45)
cv2_imshow(M45_out)
```

```python
M135 = np.array([[2,-1,-1], [-1,2,-1], [-1,-1,2]])
```

```python
M135_out = cv2.filter2D(img, -1, M135)
cv2_imshow(M135)
```

```python
x=np.array([[-1,-1,-1], [-1,8,-1], [-1, -1, -1]])
y=np.array([[-1,2,-1], [-1,2,-1], [-1,2,-1]])
```

```python
gx=cv2.filter2D(img, -1, x)
gy=cv2.filter2D(img, -1, y)
```

```python
gx
```

```python
gy
```

```python
final = cv2.addWeighted(gx,0.5,gy,0.5,0)
```

```python
cv2_imshow(final)
```

```python
#sobel operator
x = np.array([[-1,-2,-1], [0,0,0], [1,2,1]])
y = np.array([[-1,0,-1], [-2,0,2], [-1,0,1]])
```

```python
gy = cv2.filter2D(img, -1, y)
gx = cv2.filter2D(img, -1, x)
```

```python
final1 = cv2.addWeighted(gx,0.5,gy,0.5,0)
```

```python
cv2_imshow(final1)
```

```python
gx = cv2.Sobel(img, cv2.CV_64F, 1, 0)
gy = cv2.Sobel(img, cv2.CV_64F, 0, 1)
```

```python
gx = cv2.convertScaleAbs(gx)
gy = cv2.convertScaleAbs(gy)
```

```python
final2 = cv2.addWeighted(gx,0.5,gy,0.5,0)
```

```python
cv2_imshow(final2)
```
