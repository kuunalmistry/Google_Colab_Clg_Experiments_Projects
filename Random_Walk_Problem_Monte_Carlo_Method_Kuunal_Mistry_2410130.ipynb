```python
import random

# Number of simulations
num_simulations = 10000

# Number of steps in each random walk
steps = 100

# Target position
target = 10

success = 0

for _ in range(num_simulations):
    position = 0

    for _ in range(steps):
        # Randomly move +1 or -1
        position += random.choice([-1, 1])

    # Check if the target position was reached or exceeded
    if position >= target:
        success += 1

# Monte Carlo estimated probability
probability = success / num_simulations

print("Number of simulations:", num_simulations)
print("Number of steps:", steps)
print("Target position:", target)
print("Successful walks:", success)
print("Estimated probability:", probability)
```

```python
import random
import matplotlib.pyplot as plt

steps = 100

position = 0
positions = [position]

for _ in range(steps):
    position += random.choice([-1, 1])
    positions.append(position)

plt.plot(positions)
plt.axhline(0)
plt.xlabel("Number of Steps")
plt.ylabel("Position")
plt.title("Random Walk using Monte Carlo Method")
plt.show()
```
