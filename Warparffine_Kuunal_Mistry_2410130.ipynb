```python
# Image Rotation
rmatrix = cv2.getRotationMatrix2D((cols/2, rows/2), 90, 1)
rotated_img = cv2.warpAffine(img, rmatrix, (cols, rows))
cv2_imshow(rotated_img)
```

```python
ring = cv2.warpAffine(img,rmatrix,(cols, rows))
cv2_imshow(ring)
```

```python
rmatrix1 = cv2.getRotationMatrix2D((cols/2, rows/2), 180, 1)
rimg1 = cv2.warpAffine(img, rmatrix1, (cols, rows))
cv2_imshow(rimg1)
```

```python
rest1 = np.hstack((img, rimg1))
cv2_imshow(rest1)
```

```python
pt1 = np.array([[50,50], [200,50], [50,200]], dtype=np.float32)
pt2 = np.array([[10,100], [200,50], [100,250]], dtype=np.float32)
amatrix = cv2.getAffineTransform(pt1, pt2)
op = cv2.warpAffine(img, amatrix, (cols, rows))
cv2_imshow(op)
```
