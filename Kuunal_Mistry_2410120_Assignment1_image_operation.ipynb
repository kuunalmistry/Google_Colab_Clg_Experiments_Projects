```python
bit7 = ((img >> 7) & 1). astype(np.uint8)
bit6 = ((img >> 6) & 1). astype(np.uint8)
bit76 = (bit7 << 7) | (bit6 << 6)
print("Bit 7")
cv2_imshow(bit7 * 255)
print ("Bit 6")
cv2_imshow(bit6 * 255)
print("7 and 6:")
cv2_imshow(bit76)
```

```python
c = 255 / np.log(1+ np.max(img).astype(np.float64))
```

```python
log_transform = c * np.log(1 + img.astype(np.float64))
```

```python
log_transform = np.array(log_transform, dtype = np.uint8)
```

```python
cv2_imshow(log_transform)
```

Power Law Transformation

```python
gamma = 0.5
img_float = img/255.0
power = np.power(img_float, gamma)
power = np.uint8(power * 255)
cv2_imshow(power)
```
